# Proyecto: Operaciones con conjuntos en Python



# 1. Concepto de conjunto

Un conjunto es una colección de elementos diferentes, en la que no se permiten duplicados. En Python, podemos crear conjuntos utilizando la función `set()`.

En este proyecto utilizaremos nombres de pilotos de Fórmula 1 para demostrar cómo Python elimina los elementos repetidos y cómo el orden de los datos no afecta al conjunto resultante.

```python

#1: Concepto de conjunto

pilotos = [
    "Max Verstappen",
    "Lando Norris",
    "Oscar Piastri",
    "Max Verstappen",
    "Lando Norris"
]

A = set(pilotos)

print("Lista original:", pilotos)
print("Conjunto A:", A)

B = set([
    "Oscar Piastri",
    "Lando Norris",
    "Max Verstappen"
])

print("Conjunto B:", B)

print("¿A y B son iguales?:", A == B)

print("Cantidad de registros originales:", len(pilotos))
print("Cantidad de elementos únicos:", len(A))
```

# 2. Notación y pertenencia

La pertenencia indica si un elemento forma parte de un conjunto. En matemáticas se utiliza el símbolo ∈ para indicar que un elemento pertenece a un conjunto y ∉ para indicar que no pertenece.

En Python, podemos comprobar estas condiciones utilizando `in` y `not in`. También utilizaremos un conjunto universal llamado U, que contiene todos los elementos considerados en este ejemplo, y un conjunto A que contiene algunos de ellos.

```python

#2: Notacion y pertenencia

U = {
    "Max Verstappen",
    "Lando Norris",
    "Oscar Piastri",
    "George Russell",
    "Charles Leclerc"
}

A = {
    "Max Verstappen",
    "Lando Norris",
    "Oscar Piastri"
}

print("Conjunto universal U:", U)
print("Conjunto A:", A)

print("\n¿Max Verstappen pertenece a A?:",
      "Max Verstappen" in A)

print("¿George Russell no pertenece a A?:",
      "George Russell" not in A)

# Comprobar que George Russell sí pertenece a U
print("¿George Russell pertenece a U?:",
      "George Russell" in U)

print("¿A está contenido en U?:", A.issubset(U))
```

### Interpretación de los resultados

El programa demuestra que Max Verstappen pertenece al conjunto A, mientras que George Russell no pertenece a ese conjunto. Sin embargo, George Russell sí forma parte del conjunto universal U. También se comprueba que todos los elementos de A están incluidos en U. En Python, estas condiciones se verifican mediante `in`, `not in` e `issubset()`.

# 3. Extensión y comprensión de conjuntos

Un conjunto por extensión se define escribiendo directamente todos sus elementos. En cambio, un conjunto por comprensión se construye mediante una condición que determina cuáles elementos se incluyen.

En este ejemplo utilizaremos nombres de pilotos de Fórmula 1. Primero crearemos un conjunto por extensión y después construiremos otro utilizando un filtro que seleccione únicamente los pilotos que aparecen como ganadores de las carreras registradas.

```python
#3: Extension y comprension

A = {
    "Max Verstappen",
    "Lando Norris",
    "Oscar Piastri",
    "George Russell"
}

print("Conjunto por extension:", A)

import json

with open("f1_ultimas_10.json", "r", encoding="utf-8") as archivo:
    datos = json.load(archivo)

B = {
    carrera["ganador"]
    for carrera in datos["carreras"]
    if carrera["equipo"] == "McLaren"
}

print("Ganadores de McLaren por comprension:", B)
```

# 4. Conjunto vacío y conjunto universal

Un conjunto vacío es aquel que no contiene ningún elemento. En Python se crea utilizando `set()`.

El conjunto universal, representado por la letra U, contiene todos los elementos que se consideran dentro de un problema. Su contenido depende del contexto que estemos estudiando.

En este ejemplo utilizaremos pilotos de Fórmula 1 para representar un conjunto universal y un conjunto vacío.

```python

#4: Conjunto vacio y conjunto universal

vacio = set()

U = {
    "Max Verstappen",
    "Lando Norris",
    "Oscar Piastri",
    "George Russell",
    "Charles Leclerc"
}

# Mostrar los conjuntos
print("Conjunto vacio:", vacio)
print("Conjunto universal U:", U)

print("Cantidad de elementos del conjunto vacio:", len(vacio))

print("Tipo de set():", type(set()))
print("Tipo de {}:", type({}))

print("¿El conjunto esta vacio?:", len(vacio) == 0)

print("¿El conjunto vacio esta contenido en U?:",
      vacio.issubset(U))
```

# 5. Conjuntos numéricos

Los conjuntos numéricos permiten clasificar los números según sus características. Entre los principales están los naturales, enteros, racionales, reales y complejos.

- **Naturales (ℕ):** números utilizados para contar, como 1, 2, 3 y 4.
- **Enteros (ℤ):** incluyen números negativos, el cero y los positivos, como -3, 0 y 5.
- **Racionales (ℚ):** pueden expresarse como una fracción de dos enteros, con denominador diferente de cero, como 1/2 y 3/4.
- **Reales (ℝ):** incluyen los números racionales e irracionales, como 2, 1/2 y √2.
- **Complejos (ℂ):** incluyen números que pueden expresarse como a + bi, donde i² = -1.

En Python, los tipos `int`, `float` y `complex` permiten trabajar con diferentes representaciones numéricas. Sin embargo, un número `float` utiliza precisión finita, por lo que no puede representar exactamente todos los números reales.

```python

#5: Conjuntos numericos

naturales = {1, 2, 3, 4, 5}
enteros = {-3, -2, -1, 0, 1, 2, 3}
racionales = {0.5, 0.25, 1.5}
reales_ejemplo = {1, 2.5, 3.14}
complejos = {2 + 3j, 1 + 0j}

print("Naturales:", naturales)
print("Enteros:", enteros)
print("Racionales (ejemplos):", racionales)
print("Reales (ejemplos):", reales_ejemplo)
print("Complejos:", complejos)

print("\nTipo de 5:", type(5))
print("Tipo de 2.5:", type(2.5))
print("Tipo de 2 + 3j:", type(2 + 3j))

print("\n0.1 + 0.2 =", 0.1 + 0.2)
print("¿0.1 + 0.2 es exactamente igual a 0.3?:",
      0.1 + 0.2 == 0.3)
```

## 6. Subconjuntos y subconjuntos propios

Un **subconjunto** es un conjunto cuyos elementos pertenecen también a otro conjunto. En Python podemos comprobarlo con el método `issubset()` o con el operador `<=`.

Un **subconjunto propio** es un conjunto que está contenido en otro, pero tiene menos elementos. En Python se puede comprobar utilizando el operador `<`.

En este ejemplo utilizaremos nombres de pilotos de Fórmula 1 para comparar dos conjuntos.

```python

#6: Subconjuntos y subconjuntos propios

A = {"Max Verstappen", "Lando Norris"}

B = {
    "Max Verstappen",
    "Lando Norris",
    "Oscar Piastri",
    "George Russell"
}

print("Conjunto A:", A)
print("Conjunto B:", B)

print("\n¿A es subconjunto de B?:", A.issubset(B))

print("¿A es subconjunto propio de B?:", A < B)

print("¿B es subconjunto de B?:", B <= B)

print("¿B es subconjunto propio de B?:", B < B)
```

Captura de pantalla 2026-10-11 230106.png

**Interpretación de los resultados**

El conjunto A es un subconjunto de B porque todos sus elementos también se encuentran en B. Además, es un subconjunto propio porque A tiene dos elementos y B tiene cuatro.

El conjunto B también es subconjunto de sí mismo, ya que contiene todos sus elementos. Sin embargo, no es un subconjunto propio de sí mismo porque ambos conjuntos tienen exactamente los mismos elementos.

Este ejemplo demuestra cómo Python permite comparar conjuntos de manera sencilla.

## 7. Igualdad y conjuntos disjuntos

Dos conjuntos son **iguales** cuando contienen exactamente los mismos elementos, sin importar el orden en que se escriban.

Dos conjuntos son **disjuntos** cuando no tienen ningún elemento en común.

En Python podemos utilizar el operador `==` para comparar si dos conjuntos son iguales y el método `isdisjoint()` para comprobar si son disjuntos.

En este ejemplo utilizaremos nombres de pilotos de Fórmula 1.


```python

#7: Igualdad y conjuntos disjuntos

A = {"Max Verstappen", "Lando Norris", "Oscar Piastri"}

B = {"Oscar Piastri", "Max Verstappen", "Lando Norris"}

C = {"Charles Leclerc", "Lewis Hamilton"}

print("Conjunto A:", A)
print("Conjunto B:", B)
print("Conjunto C:", C)


print("\n¿A y B son iguales?:", A == B)

print("¿A y C son iguales?:", A == C)

print("¿A y B son disjuntos?:", A.isdisjoint(B))

print("¿A y C son disjuntos?:", A.isdisjoint(C))
```

**Interpretación de los resultados**

Los conjuntos A y B son iguales porque contienen los mismos tres pilotos, aunque estén escritos en diferente orden.

Los conjuntos A y C no son iguales porque tienen elementos diferentes. Además, A y B no son disjuntos porque comparten los mismos pilotos.

Por otro lado, A y C sí son disjuntos porque no tienen ningún elemento en común. Este ejemplo demuestra cómo Python permite identificar conjuntos iguales y conjuntos que no comparten elementos.

## 8. Cardinalidad de conjuntos finitos

La **cardinalidad** es la cantidad de elementos diferentes que contiene un conjunto. En Python podemos conocerla utilizando la función `len()`.

Los conjuntos no cuentan los elementos repetidos más de una vez. Por eso, si agregamos un nombre varias veces, solamente se contará una vez dentro del conjunto.

En este ejemplo utilizaremos una lista de resultados de Fórmula 1 para comparar la cantidad de registros con la cantidad de ganadores diferentes.

```python

#8: Cardinalidad de conjuntos finitos

ganadores = [
    "Max Verstappen",
    "Lando Norris",
    "Oscar Piastri",
    "Max Verstappen",
    "Lando Norris",
    "George Russell"
]

A = set(ganadores)

print("Lista original:", ganadores)
print("Conjunto de ganadores:", A)

print("\nCantidad de registros:", len(ganadores))
print("Cardinalidad del conjunto A:", len(A))

print("Registros repetidos eliminados:", len(ganadores) - len(A))
```

**Interpretación de los resultados**

La lista original contiene seis registros de ganadores, pero solamente hay cuatro nombres diferentes. Esto ocurre porque Max Verstappen y Lando Norris aparecen repetidos.

Al convertir la lista en un conjunto, Python elimina automáticamente los nombres duplicados. Por lo tanto, la cardinalidad del conjunto A es cuatro, ya que contiene cuatro elementos diferentes.

Este ejemplo demuestra que la función `len()` permite conocer la cardinalidad de un conjunto y que los elementos repetidos no aumentan su tamaño.

# 14. Diferencia simétrica

En general, nos dimos cuenta de que trabajar con conjuntos facilita mucho comparar datos y sacar conclusiones, porque evita que tengamos que revisar listas largas a mano y elimina los duplicados automáticamente.
